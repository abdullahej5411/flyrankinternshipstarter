# Week 1 Research Question

## Lane: Ranking Signal Analysis

## Research Question
Which observable content and search signals are most strongly associated with a page's visibility and performance?

## Unit of Analysis
Each row represents one content page.

## Expected Output
A signal report showing which features are most strongly associated with page performance.

## Action
Help content and SEO teams prioritize which pages to improve.

## Cost of a Wrong Recommendation
Teams may spend time improving the wrong pages while missing better opportunities.

## Why Data / ML Helps
Data and ML can identify patterns across many pages that are difficult to detect manually.

```
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Number of pages:", len(df))
print("Average CTR:", round(df["ctr"].mean(), 3))
print("Median impressions:", int(df["impressions_90d"].median()))

corr = df["avg_position"].corr(df["ctr"])
print("Correlation between avg_position and CTR:", round(corr, 3))
```

## Observation

- The dataset contains enough pages to analyze search performance signals.
- Average position and CTR show a directional relationship.
- These observations suggest that Ranking Signal Analysis is a suitable lane for further investigation.
- The findings are observational and do not imply causation.
